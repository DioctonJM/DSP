```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import step, TransferFunction

# Time range
t = np.linspace(0, 10, 1000)

# Corrected transfer functions for each damping case
systems = {
    'Underdamped (ζ=0.5, ωₙ=2)': TransferFunction([4], [1, 2, 4]),
    'Near Critically Damped (ζ≈0.9, ωₙ=1)': TransferFunction([1], [1, 1.8, 1]),
    'Heavily Underdamped (ζ=0.2, ωₙ=2)': TransferFunction([4], [1, 0.8, 4]),
    'Overdamped (ζ=2.5, ωₙ=2)': TransferFunction([4], [1, 10, 4]),
}

# Colors for each line
colors = {
    'Underdamped (ζ=0.5, ωₙ=2)': 'blue',
    'Near Critically Damped (ζ≈0.9, ωₙ=1)': 'orange',
    'Heavily Underdamped (ζ=0.2, ωₙ=2)': 'green',
    'Overdamped (ζ=2.5, ωₙ=2)': 'red',
}

# Plot the step responses
plt.figure(figsize=(12, 6))
for label, system in systems.items():
    t_out, y_out = step(system, T=t)
    plt.plot(t_out, y_out, label=label, color=colors[label])

plt.title('Step Response of 2nd Order Systems with Varying Damping')
plt.xlabel('Time (seconds)')
plt.ylabel('Response')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
